```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from itertools import combinations
from scipy.spatial.distance import pdist, squareform

def load_simplified_energy_data(file_path, state_code):
    energy_df = pd.read_excel(file_path, sheet_name='Energy_Data')
    state_data = energy_df[energy_df['State'] == state_code].copy()

    if state_data.empty:
        return None

    state_data = state_data.set_index('Year')
    state_data = state_data.rename(columns={
        'Diesel_Price': 'diesel',
        'Electricity_Price': 'elec',
        'Total_Cost': 'total price',
        'Diesel_Tax': 'diesel tax'
    })

    state_data['diesel_liter'] = (state_data['diesel'] - state_data['diesel tax']/100 - 0.244) / 3.785
    state_data['elec'] = state_data['elec'] / 100

    return state_data[['diesel', 'diesel tax', 'elec', 'total price', 'diesel_liter']]

class StateEnergyEstimator:
    def __init__(self, base_file_path, energy_file_path):
        self.data_base = pd.read_excel(base_file_path, sheet_name=None)

        self.data_harvested_area = self.data_base['harvested area']
        self.data_planted_area = self.data_base['planted area']
        self.data_production = self.data_base['production']

        self.energy_file_path = energy_file_path

        self.state_mapping = {
            0: 'AL', 1: 'AZ', 2: 'AR', 3: 'CA', 4: 'FL', 5: 'GA', 6:'KS', 7: 'LA', 8: 'MS',
            9: 'MO', 10: 'NC', 11:'NM', 12: 'OK', 13: 'SC', 14: 'TN', 15: 'TX', 16: 'VA'
        }

        plt.rcParams['xtick.direction'] = 'in'
        plt.rcParams['ytick.direction'] = 'in'
        plt.rcParams['figure.dpi'] = 300
        plt.rcParams['font.size'] = 8
        plt.rcParams['font.family'] = 'DejaVu Sans'

    def calculate_state_energy_coefficients(self, state_index):
        """Calculate energy coefficients for a specific state"""
        state_code = self.state_mapping.get(state_index, f'State_{state_index}')

        data_energy = load_simplified_energy_data(self.energy_file_path, state_code)
        if data_energy is None:
            return None, None

        state_planted_area = self.data_planted_area.iloc[state_index]
        state_production = self.data_production.iloc[state_index]

        data_total_price = data_energy['total price']
        state_total_price_per_bale = data_total_price * state_planted_area / state_production

        # Convert per bale to per 1000kg (1000kg = 4.593*480lb)
        data_energy['price per 1000kg'] = state_total_price_per_bale * 4.593

        data_energy['price per 1000kg'] = data_energy['price per 1000kg'] * 0.824  # lint allocation

        # Solve for diesel/electricity via pairwise linear equations
        coefficients = []
        for year, row in data_energy.iterrows():
            a, b, c = 1.15 * row['diesel_liter'], row['elec'], row['price per 1000kg']
            coefficients.append((a, b, c))

        return data_energy, coefficients

    def plot_energy_constraints(self, state_index, save_path=None):
        """Plot energy constraint lines for a specific state"""
        state_code = self.state_mapping.get(state_index, f'State_{state_index}')
        data_energy, coefficients = self.calculate_state_energy_coefficients(state_index)

        if data_energy is None:
            print(f"No energy data available for state {state_code}")
            return None

        x = np.linspace(0, 600, 400)

        plt.figure(figsize=(3.6, 3))

        cmap = plt.cm.get_cmap('tab20', len(data_energy.index))
        i = 8
        k = 0

        for (year, row), coef in zip(data_energy.iterrows(), coefficients):
            a, b, c = coef
            y = (c - a * x) / b
            plt.plot(x, y, label=year, linewidth=i, color=cmap.colors[k], alpha=0.7)
            i = i / 1.1
            k = k + 1

        plt.xlim(0, 500)
        plt.ylim(0, 5000)
        plt.xlabel('Diesel (L/1000kg cotton lint)')
        plt.ylabel('Electricity (kWh/1000kg cotton lint)')
        plt.legend(bbox_to_anchor=(0.52, 1), loc='upper left', fontsize=7, frameon=False, ncol=2)
        plt.title(f'Energy Constraints - {state_code}')

        if save_path:
            plt.savefig(f'{save_path}/energy_constraints_{state_code}.tiff', dpi=300, bbox_inches='tight')

        plt.show()
        return coefficients

    def find_intersection(self, coef1, coef2):
        """Find intersection point between two lines"""
        a1, b1, c1 = coef1
        a2, b2, c2 = coef2

        denominator = a1 * b2 - a2 * b1

        if denominator == 0:
            return None

        x = (c1 * b2 - c2 * b1) / denominator
        y = (c1 - a1 * x) / b1

        return (x, y)

    def calculate_intersections_and_representative_point(self, coefficients):
        """Calculate intersection points and find representative point"""
        intersections = []
        for coef1, coef2 in combinations(coefficients, 2):
            intersection = self.find_intersection(coef1, coef2)
            if intersection is not None:
                intersections.append(intersection)

        filtered_intersections = [(x, y) for x, y in intersections if 0 <= x <= 500 and 0 <= y <= 5000]

        if not filtered_intersections:
            return None, None, None

        # Find representative point (closest to all others)
        filtered_intersections = np.array(filtered_intersections)
        dist_matrix = squareform(pdist(filtered_intersections))
        average_distances = dist_matrix.sum(axis=1) / (len(filtered_intersections) - 1)
        closest_index = np.argmin(average_distances)
        closest_point = filtered_intersections[closest_index]

        return intersections, filtered_intersections, closest_point

    def plot_intersections_analysis(self, state_index, save_path=None):
        """Plot intersection analysis for a specific state"""
        state_code = self.state_mapping.get(state_index, f'State_{state_index}')
        data_energy, coefficients = self.calculate_state_energy_coefficients(state_index)

        if data_energy is None:
            print(f"No energy data available for state {state_code}")
            return None

        intersections, filtered_intersections, closest_point = self.calculate_intersections_and_representative_point(coefficients)

        if filtered_intersections is None:
            print(f"No valid intersections found for {state_code}")
            return None

        plt.figure(figsize=(3.6, 3))

        x_filtered, y_filtered = zip(*filtered_intersections)
        plt.scatter(x_filtered, y_filtered, color='#A88EC0', marker='x', s=8, zorder=3, alpha=0.4, label='Intersections')

        slope, intercept = np.polyfit(np.array(x_filtered), np.array(y_filtered), 1)

        plt.plot(np.arange(0, 500, 1), slope * np.arange(0, 500, 1) + intercept,
                linestyle='-', label=f'Fitted line\n(y={slope:.1f}x+{intercept:.1f})', color='#A4D5B1')

        # y = -4.4x + b, solve for b using representative point
        pro_elec_intercept = closest_point[1] + 4.4 * closest_point[0]

        plt.plot(np.arange(0, 500, 1), -4.4 * np.arange(0, 500, 1) + pro_elec_intercept,
                linestyle='--', label=f'Pro-electricity scenario\n(y=-4.4x+{pro_elec_intercept:.1f})', color='#FBBC04')

        fill_end_x = min(130, (pro_elec_intercept - intercept) / (slope + 4.4)) if (slope + 4.4) != 0 else 130
        fill_end_x = max(0, fill_end_x)  # Ensure non-negative

        if fill_end_x > 0:
            plt.fill_between(np.arange(0, fill_end_x, 1),
                            -4.4 * np.arange(0, fill_end_x, 1) + pro_elec_intercept,
                            slope * np.arange(0, fill_end_x, 1) + intercept,
                            color='#A4D5B1', alpha=0.5, label='Feasible area')

        # Electricity lowerbound point (y-intercept of pro-electricity line)
        plt.scatter(0, pro_elec_intercept, color='#FBBC04', marker='>', s=40, label='Theoretical electricity\nconsumption')

        plt.scatter(closest_point[0], closest_point[1], color='#EA4335', s=25, label='Representative')
        plt.annotate(f'({closest_point[0]:.1f}, {closest_point[1]:.1f})',
                    (closest_point[0], closest_point[1]), textcoords="offset points",
                    xytext=(0, -10), ha='center', fontsize=6)

        plt.xlim(0, 500)
        plt.ylim(0, 5000)
        plt.xlabel('Diesel (L/1000kg cotton lint)')
        plt.ylabel('Electricity (kWh/1000kg cotton lint)')
        plt.legend(bbox_to_anchor=(0.4, 1), loc='upper left', frameon=False, ncol=1)
        plt.title(f'Energy Optimization - {state_code}')

        if save_path:
            plt.savefig(f'{save_path}/intersections_{state_code}.tiff', dpi=300, bbox_inches='tight')

        plt.show()

        return {
            'representative_point': closest_point,
            'slope': slope,
            'intercept': intercept,
            'intersections': intersections,
            'filtered_intersections': filtered_intersections
        }

    def analyze_state(self, state_index, save_path=None):
        """Complete energy analysis for a single state"""
        state_code = self.state_mapping.get(state_index, f'State_{state_index}')
        print(f"\n🔍 Analyzing {state_code}...")

        coefficients = self.plot_energy_constraints(state_index, save_path)
        if coefficients is None:
            print(f"❌ No energy data for {state_code}")
            return None

        intersection_results = self.plot_intersections_analysis(state_index, save_path)
        if intersection_results is None:
            print(f"❌ No valid intersections for {state_code}")
            return None

        results = {
            'state_code': state_code,
            'coefficients': coefficients,
            'intersection_results': intersection_results
        }

        print(f"✅ Analysis complete for {state_code}")
        print(f"   Representative point: ({intersection_results['representative_point'][0]:.1f}, {intersection_results['representative_point'][1]:.1f})")

        return results

    def analyze_all_states(self, save_path=None):
        """Analyze energy optimization for all available states"""
        all_results = {}

        for state_index in self.state_mapping.keys():
            results = self.analyze_state(state_index, save_path)
            if results:
                all_results[state_index] = results

        return all_results

    def generate_summary_table(self, all_results):
        """Generate summary table of energy analysis results"""
        summary_data = []

        for state_index, results in all_results.items():
            state_code = results['state_code']
            rep_point = results['intersection_results']['representative_point']
            slope = results['intersection_results']['slope']
            intercept = results['intersection_results']['intercept']

            pro_elec_intercept = rep_point[1] + 4.4 * rep_point[0]
            theoretical_elec = pro_elec_intercept  # At x=0

            summary_data.append({
                'State': state_code,
                'Diesel (L/1000kg)': round(rep_point[0], 1),
                'Electricity (kWh/1000kg)': round(rep_point[1], 1),
                'Current Rate': round(slope, 1)
            })

        summary_df = pd.DataFrame(summary_data)
        return summary_df

if __name__ == "__main__":
    import warnings
    warnings.filterwarnings("ignore", message="findfont: Generic family.*not found")

    # File paths
    base_file = '../results/processed_data and results/PROCESS_GE2000_BASE.xlsx'
    energy_file = '../results/processed_data and results/Process_reorganize_Energy_state_level.xlsx'
    save_path = '../outputs/energy_state'

    estimator = StateEnergyEstimator(base_file, energy_file)

    all_results = estimator.analyze_all_states(save_path)

    summary_table = estimator.generate_summary_table(all_results)
    print("\n📊 Summary Table:")
    print(summary_table)

    summary_table.to_excel('../results/processed_data and results/Energy_Analysis_Summary.xlsx', index=False)

    print("\n🎉 State-level energy analysis complete!")
```
